# Load embeddings

Here we supply the generated embeddings of CellHermes for the Perturb-FISH dataset: ./CellHermes-Embedding/CellHermes_embbding_Perturb-FISH.pkl

```python
import scanpy as sc
import joblib
```

```python
path = "./CellHermes-Embedding/CellHermes_embbding_Perturb-FISH.pkl"
go_dict = joblib.load(path)  
```

```python
go_dict
```

# Annotation

```python
adata = sc.read("./Perturb-FISH-rawadata/Measured.h5ad")
```

```python
import numpy as np

perturb_labels = adata.obs['perturb'].astype(str).values
n_cells = adata.n_obs
n_genes = len(next(iter(go_dict.values())))

perturb_embedding = np.vstack([go_dict[label] for label in perturb_labels])

```

```python
adata.obsm['perturb_gene'] = perturb_embedding
adata.obsm['perturb_gene'].shape
```

```python
adata.write_h5ad('./Perturb-FISH-rawadata/Measured_annotated.h5ad')
```
